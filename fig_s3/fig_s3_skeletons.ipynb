```python
import pandas as pd
import matplotlib.pyplot as plt
import pymaid
import navis as nv
```

```python
#connect your catmaid instance
instance=pymaid.CatmaidInstance('https://radagast.hms.harvard.edu/catmaidaedes',"")
```

```python
def find_ribbon(tag_list):
    ribbon_count = [tag == 'ribbon synapse' for tag in tag_list]
    return sum(ribbon_count) > 0
```

```python
#MD1
MD1OSNs = pymaid.get_skids_by_annotation(['innervates MD1', 'left palp nerve', 'sensory neuron', 'PSPs done'], allow_partial = False, intersect = True)
MD1OSNsALL = pymaid.get_skids_by_annotation(['innervates MD1', 'left palp nerve', 'sensory neuron'], allow_partial = False, intersect = True)
MD1neurons = pymaid.get_neuron(MD1OSNs)
MD1neuronsALL = pymaid.get_neuron(MD1OSNsALL)
```

```python
def synapse_type_match_color(synapse_type):
    match synapse_type:
        case 'recurrent':
            return "#2500f8"#"#008bf8"
        case 'feedforward':
            return "#ff9100"#'#f5b700'
        case 'both':
            return "#33ff00"#"#04e762"
        case 'ribbon':
            return 'red'

def synapse_type_match_alpha(synapse_type):
    match synapse_type:
        case 'recurrent':
            return 0.75
        case 'feedforward':
            return 0.75
        case 'both':
            return 0.75
        case 'ribbon':
            return 0.75        
        
def synapse_type_match_size(synapse_type):
    match synapse_type:
        case 'recurrent':
            return 3
        case 'feedforward':
            return 3
        case 'both':
            return 3  
        case 'ribbon':
            return 3           
```

```python
md1_branch_lengths = []
md1_branch_order_count = []
for neuron in MD1neurons:
    connections = pymaid.get_connectors(neuron)
    recurrent_connections = pymaid.get_connectors_between(neuron,MD1neuronsALL[0:45],directional=True)
    feedforward_connections = pymaid.get_connectors_between(neuron,pymaid.get_neuron('295'),directional=True)
    both_connections = connections.loc[(connections['connector_id'].isin(recurrent_connections['connector_id'])) & (connections['connector_id'].isin(feedforward_connections['connector_id']))]
    both_connections['connection_type'] = 'both'
    recurrent_connections = connections.loc[connections['connector_id'].isin(recurrent_connections['connector_id']) & ~(connections['connector_id'].isin(feedforward_connections['connector_id'])),:]
    recurrent_connections['connection_type'] = 'recurrent'
    feedforward_connections = connections.loc[connections['connector_id'].isin(feedforward_connections['connector_id']) & ~(connections['connector_id'].isin(recurrent_connections['connector_id'])),:]
    feedforward_connections['connection_type'] = 'feedforward'
    tagged_connections = connections[~connections['tags'].isna()]
    ribbon_connections = tagged_connections.loc[[tag[0].endswith('synapse') for tag in tagged_connections['tags']]]
    ribbon_connections['connection_type'] = 'ribbon'

    all_connections = pd.concat([recurrent_connections,feedforward_connections,both_connections,ribbon_connections],ignore_index=True)

    all_connections['color']=[synapse_type_match_color(connection) for connection in all_connections['connection_type']]
    all_connections['alpha']=[synapse_type_match_alpha(connection) for connection in all_connections['connection_type']]
    all_connections['plot_size']=[synapse_type_match_size(connection) for connection in all_connections['connection_type']]

    fig, ax = nv.plot2d([neuron,all_connections[['x','y','z']]],color="black", 
                        scatter_kws={'s': 10,#all_connections['plot_size'].values,
                                    'alpha': 1, 
                                    'color': all_connections['color'].values,
                                    'edgecolor': 'black',
                                    'linewidth': .1
                                    },
                        figsize=(2.5,2.5))

    ax.azim = 270
    ax.elev = 260
    x_min, x_max = plt.gca().get_xlim()

    y_min, y_max = plt.gca().get_ylim()

    plt.savefig(fname="md1_neuron"+neuron.name+".png")
    plt.show()


```
